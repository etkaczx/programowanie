# Dekoratory w pythonie

```python
def my_decorator(func):
    def wrapper():
        my_character = '\u2740'
        print(my_character * 6)
        func()
        print(my_character * 6)

    return wrapper
```

```python
@my_decorator 
def hello():
    print('kwiatki')
    
```

```python
hello()
```

# Funkcje z argumentami

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        my_character = '\u2740'
        print(my_character * 6)d
        result = func(*args, **kwargs)
        print(my_character * 6)
        return result

    return wrapper
        
        
        
```

```python
@my_decorator
def add(a,b):
    """dodaje dwie liczby :o"""
    return a+b
```
